```python
import sys
sys.path.append("..")

import os, time
import matplotlib.pyplot as plt
import pandas as pd
from src.train import train_models
from src.evaluate import evaluate_models

# 1. Run your train and evaluate functions
pipelines, summary, (X_train, X_test, y_train, y_test) = train_models("../data/used_cars_clean_fe.csv")
metrics_df = evaluate_models(pipelines, X_train, X_test, y_train, y_test)

# 2. Record Prediction Time and merge Training Time from summary
pred_times = {}
for name, pipe in pipelines.items():
    t0 = time.time()
    pipe.predict(X_test)
    pred_times[name] = round(time.time() - t0, 4)

df = metrics_df.merge(summary, on="Model")
df["Pred Time (s)"] = df["Model"].map(pred_times)

# 3. Save merged results to CSV
os.makedirs("../outputs/metrics", exist_ok=True)
df.to_csv("../outputs/metrics/comparison.csv", index=False)

# 4. Plot Comparison Grid
os.makedirs("../outputs/figures", exist_ok=True)
plt.figure(figsize=(12, 8))

plots = [
    ("Test R2", "R² Score"),
    ("Test RMSE ($)", "RMSE ($)"),
    ("Test MAE ($)", "MAE ($)"),
    ("Train Time (s)", "Training Time (s)")
]

for i, (col, title) in enumerate(plots):
    plt.subplot(2, 2, i + 1)
    df_sorted = df.sort_values(by=col, ascending=(col != "Test R2"))
    plt.barh(df_sorted["Model"], df_sorted[col], color="steelblue")
    plt.title(title)

plt.tight_layout()
plt.savefig("../outputs/figures/model_comparison.png")
plt.show()

# Display the merged table
df
```
